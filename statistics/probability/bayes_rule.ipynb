# bayes rule
Bayes' Rule is a fundamental theorem in probability theory. It describes the probability of an event, based on prior knowledge of conditions that might be related to the event. The formula is:

$ P(A|B) = \frac{P(B|A) \times P(A)}{P(B)} $

- $P(A|B)$ is the **posterior probability**: Probability of A given B is true.
- $P(B|A)$ is the **likelihood**: Probability of B given A is true.
- $P(A)$   is the **prior probability**: Initial probability of A.
- $P(B)$   is the **marginal probability**: Total probability of B.

#### Python Example: Bayes' Rule

Let's consider a simple example. Suppose we have a bag containing 3 red and 2 blue balls. We are going to find the probability of drawing a red ball given that the previously drawn ball was also red.


```python
# Bayes Rule
# covid example

"""
in a township, 1% of people have COVID. 
a new test for detecting COVID has been devised and it correctly detects COVID 80% of the time, 
but lags non-COVID cases as COVID 10% of the time. 
given that a person tests positive, what is the probability that they have COVID?
"""
p_covid = 0.01 # 1%
p_not_covid = 1 - p_covid # 99%
p_pos_given_covid = 0.8 # 80%
p_pos_given_not_covid = 0.1 # 10%

# probability that the test will return positive, covid or not.
p_pos_test = (p_pos_given_covid * p_covid + p_pos_given_not_covid * p_not_covid)

# probability that a person has covid given that they test positive
# p(covid|pos) = p(pos|covid) * p(covid) / p(pos)
p_covid_given_pos = p_pos_given_covid * p_covid / p_pos_test

print('given values: ')
print('--------------------------------------')
print('P(covid):                     ', round(p_covid, 2))
print('P(not covid):                 ', round(p_not_covid, 2))
print('P(positive test|covid):       ', round(p_pos_given_covid, 2))
print('P(positive test|not covid):   ', round(p_pos_given_not_covid, 2))
print('')
print('calculated values: ')
print('--------------------------------------')
print('P(positive test result):      ', round(p_pos_test, 3))
print('')
print('answer: ')
print('--------------------------------------')
print('P(covid|positive test result):', round(p_covid_given_pos, 3))
```

# Naive Bayes

The Naive Bayes algorithm is a classification technique based on Bayes' Rule along with the assumption of independence among predictors. In simple terms, it assumes that the presence of a particular feature in a class is unrelated to the presence of any other feature. It's widely used in machine learning for text classification, spam filtering, and other areas.

#### Python Example: Naive Bayes Classifier

We'll use the famous Iris dataset to demonstrate the Naive Bayes classifier.

In this example, we used `GaussianNB`, a variant of Naive Bayes suited for continuous data. It assumes that features follow a normal distribution.

Both Bayes' Rule and the Naive Bayes algorithm are powerful yet straightforward tools in probability and machine learning. Their simplicity and effectiveness make them invaluable for many applications.

```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# Load Iris dataset
iris = datasets.load_iris()
X = iris.data
y = iris.target

# Splitting dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Initializing the classifier
gnb = GaussianNB()

# Training the classifier
gnb.fit(X_train, y_train)

# Making predictions
y_pred = gnb.predict(X_test)

# Calculating accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy of Naive Bayes Classifier: {accuracy}")
```
