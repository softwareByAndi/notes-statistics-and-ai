# Chebyshev's inequality 
Chebyshev's inequality is a statistical theorem that provides a bound or limit on the probability that a random variable deviates from its mean. It's a fundamental concept in probability theory and statistics.
- Chebyshev's inequality states that for any real number k>0, the probability that a random variable X deviates from its mean by more than k standard deviations is at most 1/k^2.

#### equation:
```python 
probability( abs(x - mean) >= num_std_devs * std_dev ) <= 1 / num_std_devs ** 2 
```

#### code example:
##### example 1:

````python
# chbyschev's inequality

# Assume we have a normal distribution with mean = 0 and standard deviation = 1
mean = 0
std_dev = 1
# We want to know the probability that a data point is more than 2 standard deviations away from the mean
k = 2
# Using Chebyshev's inequality
probability = 1 / (k**2)
print(f"The probability that a data point is more than {k} standard deviations away from the mean is at most {probability}")
````
