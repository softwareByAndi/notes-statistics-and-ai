# Mode

Most frequent value in a data set.  

### code example

```python
from collections import Counter
values = [1, 1, 2, 3, 3, 3, 3, 4, 5, 5, 6, 6]

# count number of occurrences of each value
value_counts = Counter(values)

# find the highest occurrence count
most_common_values = value_counts.most_common(1)
highest_count = most_common_values[0][1]

# find all values with the highest occurrence count
top_values = []
for value, count in most_common_values:
    if count == highest_count:
        top_values.append(value)
        
# output results
for value, count in value_counts.items():
    print(value, '=', count, 'times')
print('')
print('highest count =', highest_count, 'occurrences')
print('')
print('modes =', top_values)
```
