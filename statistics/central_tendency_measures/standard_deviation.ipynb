# Standard Deviation

```python
# standard deviation

sample = [9, 10.5, 11, 11, 11.3, 12, 12, 14, 14, 18]
n = len(sample)
mean = sum(sample) / n
sqrd_dist_from_mean = [(x - mean) ** 2 for x in sample]
total_sum_squares = sum(sqrd_dist_from_mean)
variance = total_sum_squares / ( n-1 )
"""
IMORTANT : var = tss/n-1 not tss/n
n-1 has something to do with the fact that the mean is an estimate. not the known average.
if the mean is not estimated and is derived from the full population, then divide by n instead of n-1
"""
standard_deviation = variance ** (1/2) # sqrt of variance because we squared the values in the beginning.

print('sample: ', sample)
print('n:      ', n)
print('mean:   ', mean)
print('sqrd_dist_from_mean:               ', [round(x, 2) for x in sqrd_dist_from_mean])
print('total_sum_squares:                 ', round(total_sum_squares, 2))
print('variance: (avg of squares)         ', round(variance, 2))
print('standard_deviation: sqrt(variance) ', round(standard_deviation, 2))
```
