## chi-square test

Is a statistical test used to compare observed results with expected results. 
- It is used to determine whether there is a significant difference between the expected frequencies and the observed frequencies in one or more categories.
- the higher the chi-square value, the more likely the difference between the observed and expected values is due to something other than chance.
- a chi-square value of 0 indicates that there is no difference between the observed and expected values.

## equation:
`chi_square = sum( (observed - expected)^2 / expected )`


#### use cases
- It is used to test the independence of two events.
- It is used to test the goodness of fit.
- It is used to test the homogeneity of a population.

## code example:

```python
# chi square

observed_values = [18, 21, 16, 5, 20, 15]
expected_values = [22, 19, 16, 11, 10, 13]

def chi_square_element(observed, expected):
    diff = observed - expected
    squared_diff = diff**2
    normalized = squared_diff / expected
    return normalized

chi_square_elements = [chi_square_element(observed, expected) for observed, expected in zip(observed_values, expected_values)]
chi_square = sum(chi_square_elements)
print('observed_values:     ', observed_values)
print('expected_values:     ', expected_values)
print('chi_square_elements: ', chi_square_elements)
print('chi_square:          ', chi_square)
```
