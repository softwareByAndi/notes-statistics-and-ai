## correlation coefficient
Is a measure of how strongly correlated two variables are. 
- It is denoted by the symbol r and takes values between -1 and +1. 
- A positive value for r indicates a positive association, and a negative value for r indicates a negative association. 
- The closer r is to 1 or -1, the stronger the association between the two variables. 
- If r is close to 0, it means there is no relationship between the variables.


<font size=7>

$ r = \frac{\sum ((x_i - \overline{x}) \cdot (y_i - \overline{y}))}{(n-1) \cdot \sigma_x \cdot \sigma_y} $

</font>

### code example:

```python
# data set

data_x = [32, 45, 39, 43, 58, 84, 65]
data_y = [17, 20, 23, 7, 24, 49, 38]
n = len(data_x)

mean_x = sum(data_x) / n
mean_y = sum(data_y) / n

# IMPORTANT: use n-1 for sample standard deviation
std_dev_x = (sum([(x - mean_x)**2 for x in data_x]) / (n-1))**0.5 
std_dev_y = (sum([(y - mean_y)**2 for y in data_y]) / (n-1))**0.5

print('data set:')
print('x:         ', data_x)
print('y:         ', data_y)
print('n:         ', n)
print('mean_x:    ', round(mean_x, 2))
print('mean_y:    ', round(mean_y, 2))
print('std_dev_x: ', round(std_dev_x, 2))
print('std_dev_y: ', round(std_dev_y, 2))

# plot data points
import matplotlib.pyplot as plt
plt.scatter(data_x, data_y)
plt.show()
```

<br>

#### equation 1:

<font size=5>

$ r = \frac{\sum ((x_i - \overline{x}) \cdot (y_i - \overline{y}))}{(n-1) \cdot \sigma_x \cdot \sigma_y} $

</font>

```python
# calculate correlation coefficient
# equation 1: r = sum((x - mean_x) * (y - mean_y)) / ((n-1) * std_dev_x * std_dev_y)

diffs_x = [x - mean_x for x in data_x]
diffs_y= [y - mean_y for y in data_y]

prods_dx_dy = [dx * dy for dx, dy in zip(diffs_x, diffs_y)]
sum_diff_prods = sum(prods_dx_dy)

denominator = (n-1) * std_dev_x * std_dev_y
correlation = sum_diff_prods / denominator

# print
print('data:           ', data_x, '  :  ', data_y)
print('means :         ', round(mean_x, 2), ' : ', round(mean_y, 2))
print('std_devs:       ', round(std_dev_x, 2), ' : ', round(std_dev_y, 2))
print('diffs_x:        ', [round(i, 2) for i in diffs_x], '  :  ', [round(i, 2) for i in diffs_y])
print('prods_dx_dy:    ', [round(i, 2) for i in prods_dx_dy])
print('sum_diff_prods: ', round(sum_diff_prods, 2))
print('denominator:    ', round(denominator, 2))
print('correlation:    ', round(correlation, 2))
```

<br>

#### equation 2:

<font size=5>

$ r = \frac{\sum (x_i \cdot y_i) - n \cdot \overline{x} \cdot \overline{y}}{(n-1) \cdot \sigma_x \cdot \sigma_y} $

</font>

```python
# caluclate correlation coefficient
# equation 2: r = sum(x * y) - n * mean_x * mean_y / ((n-1) * std_dev_x * std_dev_y)

prods_data_x_y = [x * y for x, y in zip(data_x, data_y)]

numerator   = sum(prods_data_x_y) - (n * mean_x * mean_y)
denominator = (n-1) * std_dev_x * std_dev_y
correlation = numerator / denominator

# print
print('data:           ', data_x, '  : ', data_y)
print('means:          ', round(mean_x, 2), ' : ', round(mean_y, 2))
print('std_devs:       ', round(std_dev_x, 2), ' : ', round(std_dev_y, 2))
print('prods_data_x_y: ', prods_data_x_y)
print('numerator:      ', round(numerator, 2))
print('denominator:    ', round(denominator, 2))
print('correlation:    ', round(correlation, 2))
```
