```python
import pandas as pd
pd.set_option('display.max_rows', 500)
pd.set_option('display.max_columns', 500)
pd.set_option('display.width', 1000)
def display(header, item):
    print(header, ':')
    print(item, '\n')
```

# pd.Series

```python
med_names = ['omeprazole', 'azithromycin', 'metformin', 'ibuprofen', 'cetirizine']
old_price_list = [75, 50, 90, 40, 70]
new_price_list = [80, 60, 100, 45, 80]

old_price_list_labeled = pd.Series(old_price_list, index=med_names)
new_price_list_labeled = pd.Series(new_price_list, index=med_names).sort_index()

print(old_price_list_labeled, '\n')
print(new_price_list_labeled, '\n')
print(new_price_list_labeled - old_price_list_labeled)
```

```python
days = pd.Series([420, 380, 390], index=['day1', 'day2', 'day3'])
print(days[:'day3'], '\n') # inclusive indexing
print(days[0:2], '\n')     # exclusive indexing
```

# pd.DataFrame

see [quick_ref.py](./quick_ref.py)

```python
df_test = pd.DataFrame({
    'name': ['John', 'Jane', 'Joe', 'Jill', 'Jack'],
    'age': [23, 25, 21, 22, 24],
    'city': ['Los Angeles', 'San Diego', 'Portland', 'Chicago', 'Boston']
})
df_test['net_worth'] = [20000, 100000, 100, 5000, 55000]
df_test
```

## loc & iloc

```python
df_copy = df_test.copy()

print(df_copy.loc[[1, 3], ['name', 'age']], '\n') # loc is label based

df_copy.loc[3, 'name'] = 'Jillian'
df_copy.loc[[1, 3], ['age']] = [26, 29]

print(df_copy.iloc[[1, 3], [0, 1]], '\n')         # iloc is integer based

df_copy.loc[df_copy['age'] > 23, 'city'] = 'San Francisco'

df_copy
```

## concat, merge, join

## concat

```python
new_records = pd.DataFrame({
    'name': ['Jen', 'Jeff'],
    'age': [26, 24],
    'city': ['Seattle', 'Portland'],
    'net_worth': [75000, 400]
}, index=[5, 6])

df_copy = pd.concat([df_copy, new_records])
df_copy
```

```python
display('concat over axis=1', pd.concat([df_test, df_copy], axis=1, keys=['original', 'modified']))
display('concat over axis=0', pd.concat([df_test, df_copy], axis=0, keys=['original', 'modified']))
```

## merge

```python
display("default (inner) merge on name", pd.merge(df_test, df_copy, on='name', suffixes=('_original', '_modified')))
display("outer merge on name", pd.merge(df_test, df_copy, how="outer", on='name', suffixes=('_original', '_modified')))
```

## join

```python
display("default (left) join on index", df_test.join(df_copy, lsuffix='_old', rsuffix='_new'))
display("right join on index", df_test.join(df_copy, how='right', lsuffix='_old', rsuffix='_new'))
```

# statistics

```python
display('describe', df_copy['age'].describe().round(2)) # summary statistics
```

```python
display('value_counts', df_copy['age'].value_counts(normalize=True).round(2)) # % ratio of each unique value
```

```python
print('age       - unique: ', df_copy['age'].unique())
print('age       - nunique:', df_copy['age'].nunique()) # number of unique values
print('net worth - min:    ', df_copy['net_worth'].min())
print('net worth - max:    ', df_copy['net_worth'].max())
print('net worth - mean:   ', df_copy['net_worth'].mean().round(2))
print('net worth - median: ', df_copy['net_worth'].median())
print('net worth - std:    ', df_copy['net_worth'].std().round(2))
print('net worth - sum:    ', df_copy['net_worth'].sum())

# display('cumulative sum', df_copy['age'].cumsum()) # cumulative sum
# print('idxmax:   ', df_copy['age'].idxmax()) # index of max value
# print('idxmin:   ', df_copy['age'].idxmin()) # index of min value
# print('age - prod:   ', df_copy['age'].prod())
```

## group by

```python
display(
    'mean net worth, grouped by city', 
    df_copy.groupby('city')['net_worth'].mean().round(2)
)
```

# apply

```python
def k(num):
    if num < 1000:
        return f'${num/1000:.1f}K'
    else:
        return f'${int(num/1000)}K'
display('apply', df_copy['net_worth'].apply(k))
```
