# python pandas 
## types of series

```python
import pandas as pd
from datetime import datetime

# Starting today, create a range of 10 daily dates
start_date = datetime.today().strftime('%Y-%m-%d')
date_range = pd.date_range(start=start_date, periods=10, freq='D')

# Create a DataFrame with these dates
date_df = pd.DataFrame({'Date': date_range})
date_df['date_string'] = date_df['Date'].dt.strftime('%Y/%m/%d')

# Display the Types
print(f'type start_date:               {type(start_date)}')
print(f'type date_range:               {type(date_range)}')
print('date_df.info(): ')
print(date_df.info())
print("""
-- Notice here that the dtype of the 'Date' column is datetime64[ns] and the dtype of the 'date_string' column is object, instead of string.
-- When this happens, you can get types of individual elements in a column by applying the `type` function to each element in the 'date_string' Series.
-- And then we can get the unique types of the newly created series using .unique().
""")
unique_date_string_types = (
    date_df['date_string']
    .apply(type)
    .unique()
)
print(f"unique date string types:      {unique_date_string_types}")

# Display the DataFrame
print('')
print(date_df)
```
