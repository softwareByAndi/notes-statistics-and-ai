```python
df_automobiles = pd.read_csv('../test_data/Automobile.csv')
```

# histogram (numerical data)

```python
def ax_rot_90(ax):
    ax.xaxis.set_tick_params(rotation=90) # rotate x-axis labels
```

```python
# Create a figure and a set of subplots with 1 row and 3 columns
fig, (ax0, ax1, ax2) = plt.subplots(1, 3, figsize=(15, 3))
for ax in [ax0, ax1, ax2]:
    ax_rot_90(ax)

# examples
sns.histplot(df_automobiles['price'], color='orange', ax=ax0)
sns.histplot(data=df_automobiles, x='price', kde=True, bins=20,       color='red',  ax=ax1)
sns.histplot(data=df_automobiles, x='price', kde=True, binwidth=1000, color='gray', ax=ax2)

plt.tight_layout()
plt.show()
```

```python
# Create a figure and a set of subplots with 1 row and 3 columns
fig, (ax0, ax1) = plt.subplots(1, 2, figsize=(15, 3))

# calculating binwidth
iqr = df_automobiles['price'].quantile(0.75) - df_automobiles['price'].quantile(0.25)
n = df_automobiles['price'].count()
binwidth = 2 * iqr * n ** (-1/3)

# calculating number of bins
range = df_automobiles['price'].max() - df_automobiles['price'].min()
num_bins = int(np.ceil(range / binwidth))

ax0.set_title('binwidth')
ax1.set_title('num_bins')
sns.histplot(data=df_automobiles, x='price', kde=True, color='gray', ax=ax0, binwidth=binwidth) # width of each bin
sns.histplot(data=df_automobiles, x='price', kde=True, color='red',  ax=ax1, bins=num_bins) # number of bins

plt.tight_layout()
plt.show()
```

# boxplot

```python
# Create a figure and a set of subplots with 1 row and 3 columns
fig, (ax0, ax1, ax2) = plt.subplots(1, 3, figsize=(20, 5))

ax0.set_title('curb_weight')
sns.boxplot(
    data=df_automobiles, 
    x='curb_weight', 
    color='skyblue', 
    whis=1.5,
    ax=ax0
)

ax1.set_title('body style v.s. price')
sns.boxplot(
    data=df_automobiles, 
    x='body_style', 
    y='price', 
    hue='body_style',  # hue is required to set palette
    palette='viridis', # this is not necessary, but it makes the plot look better
    whis=1.5, 
    ax=ax1
)

ax2.set_title('body style v.s. price v.s. number of doors')
sns.boxplot(
    data=df_automobiles, 
    x='body_style', 
    y='price', 
    hue='number_of_doors', # adds a third dimension to the plot
    palette='viridis', 
    whis=1.5, 
    ax=ax2
)

ax1.xaxis.set_tick_params(rotation=90) # rotate x-axis labels
ax2.xaxis.set_tick_params(rotation=90) # rotate x-axis labels
# adjusts the subplots to fit into the figure area
plt.tight_layout()
plt.show()
```

### boxplots can't see multimodal distributions

```python
# Create a figure and a set of subplots with 1 row and 2 columns
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 3))

ax1.set_title('Boxplot of Bore')
sns.boxplot(
    data=df_automobiles, 
    x='bore', 
    color='skyblue', 
    whis=1.5, 
    ax=ax1
)

ax2.set_title('Histogram of Bore')
sns.histplot(
    data=df_automobiles, 
    x='bore', 
    color='red', 
    kde=True, 
    ax=ax2
)

# adjusts the subplots to fit into the figure area
plt.tight_layout() 
plt.show()
```

# line plot

```python
# load one of the example datasets that are included with seaborn
flights = sns.load_dataset('flights')

fig, [ax0, ax1] = plt.subplots(1, 2, figsize=(15, 7))

ax0.set_title('month v.s. passengers')
ax0.set_ylabel('number of passengers')

# the shaded area is the 95% confidence interval
# the line is the mean
sns.lineplot(
    data=flights, 
    x='month', 
    y='passengers',
    # errorbar=('ci', False),
    ax=ax0
)

ax1.set_title('month v.s. passengers v.s. year')
ax1.set_ylabel('number of passengers')
sns.lineplot(
    data=flights, 
    x='month', 
    y='passengers', 
    hue='year', 
    palette='viridis'
)
ax1.legend(bbox_to_anchor=(1, 1)) # place the legend location

plt.tight_layout()
plt.show()
```
