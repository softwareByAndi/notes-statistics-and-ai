### standard deviation:

````python
# standard deviation

sample = [9, 10.5, 11, 11, 11.3, 12, 12, 14, 14, 18]
n = len(sample)
mean = sum(sample) / n
sqrd_dist_from_mean = [(x - mean) ** 2 for x in sample]
total_sum_squares = sum(sqrd_dist_from_mean)
variance = total_sum_squares / ( n-1 )
"""
IMORTANT : var = tss/n-1 not tss/n
n-1 has something to do with the fact that the mean is an estimate. not the known average.
if the mean is not estimated and is derived from the full population, then divide by n instead of n-1
"""
standard_deviation = variance ** (1/2) # sqrt of variance because we squared the values in the beginning.

print('sample: ', sample)
print('n:      ', n)
print('mean:   ', mean)
print('sqrd_dist_from_mean:               ', [round(x, 2) for x in sqrd_dist_from_mean])
print('total_sum_squares:                 ', round(total_sum_squares, 2))
print('variance: (avg of squares)         ', round(variance, 2))
print('standard_deviation: sqrt(variance) ', round(standard_deviation, 2))
````

### Relative Dispersion / coefficient of variation (CV):
CV (coefficient of variation) is a measure of relative dispersion, calculated as the ratio of the standard deviation to the mean.  

The CV is commonly used in statistics to estimate risk or compare the variability of different datasets. It provides a way to assess the relative variability of data points, regardless of their scale or units of measurement.  

- smaller ratio = more consistent
- can be used to compare 2 data that have different units of measurement
- `cv = std_dev/mean`
- also see: sharp ratio

**python example:**

````python
apples = [3, 5, 6, 4, 3, 5, 4]
oranges = [1, 3, 7, 9, 2, 6, 2]

n_a = len(apples)
n_o = len(oranges)
m_a = sum(apples) / n_a
m_o = sum(oranges) / n_o

std_a = ( sum( [(x - m_a) ** 2 for x in apples ] ) / n_a ) ** (1/2)
std_o = ( sum( [(x - m_o) ** 2 for x in oranges] ) / n_o ) ** (1/2)

cv_a = std_a / m_a
cv_o = std_o / m_o

print('         apples : oranges')
print('   mean   ', round(m_a, 2), ':', round(m_o, 2))
print('std_dev   ', round(std_a, 2), ':', round(std_o, 2))
print('     cv   ', round(cv_a, 2), ':', round(cv_o, 2))
````

# Chebyshev's inequality 
Chebyshev's inequality is a statistical theorem that provides a bound or limit on the probability that a random variable deviates from its mean. It's a fundamental concept in probability theory and statistics.
- Chebyshev's inequality states that for any real number k>0, the probability that a random variable X deviates from its mean by more than k standard deviations is at most 1/k^2.

#### equation:
```python 
probability( abs(x - mean) >= num_std_devs * std_dev ) <= 1 / num_std_devs ** 2 
```

#### code example:

````python
sample_size = 40
std_dev = 3
interval = [22, 34]
mean = 28
distances_from_mean = [ abs(x - mean) for x in interval ]
std_devs_from_mean = [ d / std_dev for d in distances_from_mean ]

# probability_of( abs(x - mean) >= num_std_devs * std_dev ) <= 1 / num_std_devs ** 2
# this equation is basically saying that the maximum probability of a value being more than N * std_dev away from the mean is 1 / N ** 2
# in other words, the real probability is can be less than that, but not more.

# this is the probability for both sides of the bell curve
def probability(num_std_devs):
    return 1 / num_std_devs ** 2

# FIXME : I think there may be a logical fallacy below. I think the probability of being outside the interval is not the same as the sum of probabilities of being outside the interval on either side...

# percentage of values outside of the interval
interval_percentiles = [
    probability(x) / 2 # FIXME logical falacy here? -- p(x)/2 because we are only looking at one side of the interval
    for x in std_devs_from_mean
]
max_probability_outside_interval = sum(interval_percentiles)
min_probability_inside_interval = 1 - max_probability_outside_interval

print('distnaces_from_mean:          ', distances_from_mean)
print('std_devs_from_mean:           ', std_devs_from_mean)
print('interval_percentiles:         ', interval_percentiles)
print('maximum probability of being outside: ', max_probability_outside_interval)
print('minimum probability of being inside:  ', min_probability_inside_interval)
````
